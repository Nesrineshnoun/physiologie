### Étape 2 : Visualiser le contenu du fichier ZIP
Nous allons lister tous les fichiers contenus dans l'archive sans l'extraire entièrement.

```python
import zipfile
import os

zip_path = '/content/drive/MyDrive/Frontiers-Machine Learning.zip'

if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        # Liste des fichiers
        file_list = zip_ref.namelist()
        print(f"Nombre de fichiers dans l'archive : {len(file_list)}\n")
        print("Liste des 20 premiers fichiers :")
        for file in file_list[:20]:
            print(f" - {file}")
        if len(file_list) > 20:
            print("\n... et d'autres fichiers.")
else:
    print(f"Le fichier n'a pas été trouvé au chemin : {zip_path}. Veuillez vérifier le chemin.")
```

### Étape 3 : Extraction et chargement du fichier Excel
Nous extrayons le fichier Excel `Cantor_machineclean-opendepository.xlsx` détecté dans l'archive ZIP pour pouvoir l'analyser.

```python
import zipfile
import pandas as pd

zip_path = '/content/drive/MyDrive/Frontiers-Machine Learning.zip'
extract_path = '/content/extracted_data'

# Extraction du fichier
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# Chemin vers le fichier Excel extrait
excel_file = f"{extract_path}/Frontiers-Machine Learning/Cantor_machineclean-opendepository.xlsx"

# Lecture du fichier Excel
df = pd.read_excel(excel_file)
print("Données chargées avec succès !")
print(f"Dimensions de la table : {df.shape[0]} lignes, {df.shape[1]} colonnes.")
display(df.head())
```
